# Análise de Rede: Padrões de Comunicação e Hipótese de Gargalo

**Hipótese**: a comunicação entre dispositivos é um gargalo para a latência de inferência?

Ao paralelizar a inferência de um LLM em múltiplos nós (via Tensor Parallelism ou Pipeline
Parallelism), cada passo de geração de token passa a depender de troca de dados pela rede —
ativações, no caso de PP, ou resultados parciais de matmuls, no caso de TP. Se a rede for lenta
ou os padrões de comunicação forem ineficientes, esse tráfego pode se tornar o fator limitante
da latência, e não o cômputo em si.

Este notebook:
1. Carrega e faz o parsing dos traces de telemetria de rede e latência de todos os experimentos
2. Caracteriza os padrões de comunicação (volume, burstiness/rajadas, atividade, topologia)
3. Correlaciona os sinais de comunicação com a latência para testar a hipótese de gargalo


## Configuração Inicial

Importa as bibliotecas usadas na análise, define os diretórios de entrada (traces brutos em
`data/final/trace`) e saída (CSVs processados e figuras), e fixa os parâmetros visuais para
manter a mesma tipografia dos demais notebooks (`inference_analysis.ipynb`,
`communication_analysis.ipynb`), garantindo consistência visual entre eles.


````python
import os
import re
import math
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import seaborn as sns

DATA_ROOT = Path('../data')
TRACE_ROOT = DATA_ROOT / 'final' / 'trace'
OUTPUT_DIR = Path('../data_processed')
IMAGES_DIR = Path('../figures/network_analysis')

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
IMAGES_DIR.mkdir(parents=True, exist_ok=True)

# Same typography/style as inference_analysis.ipynb, for visual consistency across notebooks.
plt.rcParams.update({
    'font.size': 13,
    'axes.titlesize': 18, 'axes.titleweight': 'bold',
    'axes.labelsize': 15,
    'xtick.labelsize': 14, 'ytick.labelsize': 13,
    'legend.fontsize': 13, 'legend.title_fontsize': 14,
    'figure.titlesize': 20, 'figure.titleweight': 'bold',
})

def bar_figsize_for(n_items):
    w = 3.0 + 0.9 * n_items
    return (min(max(w, 6.0), 18.0), 7.0)

experiment_paths = sorted([p for p in TRACE_ROOT.iterdir() if p.is_dir()])
print(f'Found {len(experiment_paths)} experiments')

# Sequence-length here comes from summary.csv's `workload` column, derived in
# data_preprocessing.ipynb from actual measured token counts — several "*_long_*"-named
# folders (see parse_exp_name below) actually ran short (128-token) workloads, so the
# folder-name suffix alone is not reliable. Fallback: sufixo do nome da pasta.
_summary = pd.read_csv(OUTPUT_DIR / 'summary.csv')
WORKLOAD_MAP = (
    _summary.set_index('experiment_name')['workload']
    if 'workload' in _summary.columns else pd.Series(dtype=object)
)

def workload_for(exp_name):
    wl = WORKLOAD_MAP.get(exp_name)
    if isinstance(wl, str):
        return wl
    return 'Long' if '_long_' in exp_name else 'Short'
````

## Funções Auxiliares

Os nomes brutos de experimentos e nós seguem uma convenção compacta usada pelos scripts de coleta
(por exemplo, `N2_poti_TP_long_r1_790072` ou `poti1`). Esta seção concentra as funções que
decodificam essa convenção e padronizam rótulos para uso em todos os gráficos do notebook.

- **`parse_exp_name`**: separa o nome bruto em campos estruturados (tamanho do cluster, cluster
  físico, estratégia de paralelismo, tag de sequência).
- **`format_node_name`**: formata nomes de nó como `POTI-1`.
- **`format_experiment_name`**: produz rótulos de duas linhas (sistema + estratégia) adequados
  para eixos de gráficos, já usando `workload_for` para corrigir o rótulo de comprimento de
  sequência quando o nome da pasta não é confiável.
- **`get_experiment_type`**: classifica o experimento em `tp`, `pp` ou `none`, usado para colorir
  consistentemente os gráficos por estratégia de paralelismo.
- **`experiment_sort_key`**: ordena experimentos por tamanho de cluster, depois por estratégia e
  por workload, dando uma ordem visual estável em todas as figuras.
- **`trim_idle_tails_and_normalize_time`**: recorta as pontas ociosas do trace (antes do início e
  depois do fim da atividade de rede) e normaliza o eixo do tempo para começar em zero, evitando
  que tempo de setup/teardown distorça as séries temporais.

Todas as funções de formatação tratam entradas inesperadas retornando o valor original em vez de
lançar exceções, já que essas funções são usadas em laços de plotagem onde uma falha silenciosa é
preferível a interromper a geração de todas as figuras.


````python
COMM_THRESHOLD = 10.0  # kB/s
TYPE_ORDER = {'tp': 0, 'pp': 1, 'none': 2}
PARALLELISM_COLORS = {'none': 'tab:orange', 'pp': 'tab:blue', 'tp': 'tab:green'}
PARALLELISM_LABELS = {'none': 'No Comm', 'pp': 'Pipeline Parallel', 'tp': 'Tensor Parallel'}

# Short/Long agora combinados na mesma figura em vez de split; cor já é usada pela
# estratégia de paralelismo (PARALLELISM_COLORS), então workload usa um canal livre:
# linestyle (linhas), marker (scatter) ou hatch (barras).
WORKLOAD_LS = {'Short': '--', 'Long': '-'}
WORKLOAD_MARKERS = {'Short': 'o', 'Long': 's'}
WORKLOAD_HATCH = {'Short': '', 'Long': '///'}
WORKLOAD_HATCH_PATCHES = [
    Patch(facecolor='white', edgecolor='black', hatch=h, label=wl)
    for wl, h in WORKLOAD_HATCH.items()
]


def parse_exp_name(name):
    parts = name.split('_')
    return {
        'exp_name': name,
        'cluster_size': int(parts[0][1:]),
        'cluster': parts[1],
        'parallelism': parts[2],
        'seqlen': parts[3],
    }


def format_node_name(node):
    if not isinstance(node, str):
        return str(node).upper()
    match = re.match(r'([a-zA-Z]+)(\d+)', node)
    return f'{match.group(1).upper()}-{match.group(2)}' if match else node.upper()


def format_experiment_name(name):
    if not isinstance(name, str) or pd.isna(name):
        return 'Unknown Experiment'
    parts = name.split('_')
    if len(parts) < 4:
        return name
    try:
        n_nodes = parts[0][1:]
        system = parts[1].upper()
        strategy = parts[2]
        node_text = f'{n_nodes} Node' if n_nodes == '1' else f'{n_nodes} Nodes'
        strat_map = {'none': 'No Comm', 'pp': 'Pipeline Parallel', 'tp': 'Tensor Parallel',
                     'PP': 'Pipeline Parallel', 'TP': 'Tensor Parallel'}
        strategy_text = strat_map.get(strategy, strategy.upper())
        len_text = workload_for(name)
        return f'{node_text} {system}\n{strategy_text} ({len_text})'
    except Exception:
        return name


def get_experiment_type(exp_name):
    name = exp_name.lower()
    if 'tp' in name:
        return 'tp'
    if 'pp' in name:
        return 'pp'
    return 'none'


def parallelism_legend_patches(exp_names):
    types_present = sorted({get_experiment_type(e) for e in exp_names}, key=lambda t: TYPE_ORDER.get(t, 99))
    return [Patch(facecolor=PARALLELISM_COLORS[t], edgecolor='black', label=PARALLELISM_LABELS[t])
            for t in types_present]


def experiment_sort_key(exp_name):
    m = parse_exp_name(exp_name)
    strategy_rank = TYPE_ORDER.get(m['parallelism'].lower(), 99)
    workload_rank = {'Short': 0, 'Long': 1}.get(workload_for(exp_name), 99)
    return (m['cluster_size'], m['cluster'], strategy_rank, workload_rank)


def save_fig(fig, filename):
    fig.savefig(IMAGES_DIR / filename, dpi=150, bbox_inches='tight')
    plt.show()


def second_level_node_df(exp_df):
    return (
        exp_df.assign(timestamp=pd.to_datetime(exp_df['timestamp']))
        .groupby(['node_name', pd.Grouper(key='timestamp', freq='1s')])['total_kB_s']
        .sum()
        .reset_index()
    )


def trim_idle_tails_and_normalize_time(exp_df, threshold_kB_s=500.0, margin_seconds=10):
    activity = exp_df.groupby('timestamp')['total_kB_s'].sum().sort_index()
    active_mask = activity > threshold_kB_s
    exp_df = exp_df.copy()
    if active_mask.any():
        start_time = activity.index[active_mask][0] - pd.Timedelta(seconds=margin_seconds)
        end_time = activity.index[active_mask][-1] + pd.Timedelta(seconds=margin_seconds)
        exp_df = exp_df[
            (exp_df['timestamp'] >= start_time) & (exp_df['timestamp'] <= end_time)
        ].copy()
    exp_df['t'] = (exp_df['timestamp'] - exp_df['timestamp'].min()).dt.total_seconds()
    return exp_df


ordered_experiments = sorted([p.name for p in experiment_paths], key=experiment_sort_key)
````

## Carregamento e Normalização da Telemetria

Os arquivos `network_telemetry.csv` têm um formato de colunas separadas por espaço (semelhante à
saída do `sar -n DEV`), não um CSV convencional. A função `load_network_csv` faz o parsing linha a
linha, validando que cada linha tem exatamente 13 campos antes de extrair as métricas de interesse:
pacotes e kB/s recebidos e transmitidos (`rx_*`, `tx_*`) e a utilização da interface
(`ifutil_pct`). Linhas malformadas são descartadas silenciosamente em vez de interromper o
carregamento de um experimento inteiro.

Em seguida, percorremos todos os diretórios de experimento em busca desses arquivos (um por nó),
concatenamos tudo em `network_ts_df` e derivamos `total_kB_s = rx_kB_s + tx_kB_s`, a métrica de
throughput agregado usada no restante do notebook. O resultado é salvo em
`data_processed/network_timeseries.csv` para reuso sem precisar reprocessar os traces brutos.


````python
def find_network_files(experiment_path):
    files = []
    for root, _, filenames in os.walk(experiment_path):
        if 'network_telemetry.csv' in filenames:
            files.append(Path(root) / 'network_telemetry.csv')
    return files


def load_network_csv(path):
    rows = []
    with open(path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 13:
                continue
            rows.append({
                'timestamp': pd.to_datetime(f'{parts[0]} {parts[1]}'),
                'interface': parts[4],
                'rx_packets_s': float(parts[5]),
                'tx_packets_s': float(parts[6]),
                'rx_kB_s': float(parts[7]),
                'tx_kB_s': float(parts[8]),
                'ifutil_pct': float(parts[12]),
            })
    return pd.DataFrame(rows)


network_rows = []
for exp_path in experiment_paths:
    for net_file in find_network_files(exp_path):
        node_name = net_file.parent.name
        try:
            df = load_network_csv(net_file)
            df['experiment_name'] = exp_path.name
            df['node_name'] = node_name
            df['total_kB_s'] = df['rx_kB_s'] + df['tx_kB_s']
            network_rows.append(df)
        except Exception as e:
            print(f'Error loading {net_file}: {e}')

if network_rows:
    network_ts_df = pd.concat(network_rows, ignore_index=True)
    network_ts_df.to_csv(OUTPUT_DIR / 'network_timeseries.csv', index=False)
    print(f'Loaded {network_ts_df.shape[0]} rows from {len(experiment_paths)} experiments')
    network_ts_df.head()
else:
    print('Warning: No network_telemetry.csv files found in experiments.')
    print('Available files in first experiment:', list(experiment_paths[0].rglob('*'))[:5] if experiment_paths else 'No experiments found')
````

````python
summary_rows = []
for (experiment_name, node_name), group in network_ts_df.groupby(['experiment_name', 'node_name']):
    summary_rows.append({
        'experiment_name': experiment_name,
        'node_name': node_name,
        'rx_avg_kB_s': group['rx_kB_s'].mean(),
        'tx_avg_kB_s': group['tx_kB_s'].mean(),
        'total_avg_kB_s': group['total_kB_s'].mean(),
        'rx_peak_kB_s': group['rx_kB_s'].max(),
        'tx_peak_kB_s': group['tx_kB_s'].max(),
        'total_peak_kB_s': group['total_kB_s'].max(),
        'ifutil_avg_pct': group['ifutil_pct'].mean(),
        'ifutil_peak_pct': group['ifutil_pct'].max(),
    })
network_node_summary_df = pd.DataFrame(summary_rows)
network_node_summary_df.head()
````

## Carregamento das Métricas de Latência

Extrai TTFT (*Time To First Token*) e latência entre tokens (*inter-token latency*) do arquivo
`profile_export_aiperf.csv`, gerado pelo `aiperf` apenas no nó de cabeça (*head node*), já que é
esse nó que recebe as requisições e mede a latência ponta a ponta observada pelo cliente.

`extract_latency_metrics` faz um parsing tolerante do CSV: procura, linha a linha, por métricas
cujo nome contenha "ttft"/"time to first token" ou "inter" + "token", lendo a coluna de média
(índice 1) e, quando presente, a coluna de p99 (índice 13) do relatório do `aiperf`.
`aggregate_latency_signals` percorre os diretórios de nó de um experimento e retorna o primeiro
arquivo de perfil encontrado — só o nó de cabeça deve ter esse arquivo, então isso equivale a
selecioná-lo diretamente.


````python
def extract_latency_metrics(csv_path):
    metrics = {}
    try:
        with open(csv_path, 'r') as f:
            for line in f:
                line = line.strip()
                if not line or ',' not in line:
                    continue
                parts = [p.strip() for p in line.split(',')]
                if len(parts) < 2:
                    continue
                metric_name = parts[0].lower()
                if 'ttft' in metric_name or 'time to first token' in metric_name:
                    try:
                        metrics['TTFT_avg_ms'] = float(parts[1])
                    except Exception:
                        pass
                    if len(parts) > 13:
                        try:
                            metrics['TTFT_p99_ms'] = float(parts[13])
                        except Exception:
                            pass
                if 'inter' in metric_name and 'token' in metric_name:
                    try:
                        metrics['inter_token_avg_ms'] = float(parts[1])
                    except Exception:
                        pass
                    if len(parts) > 13:
                        try:
                            metrics['inter_token_p99_ms'] = float(parts[13])
                        except Exception:
                            pass
    except Exception:
        pass
    return metrics


def aggregate_latency_signals(exp_path):
    for node_dir in sorted(exp_path.iterdir()):
        if not node_dir.is_dir():
            continue
        profile_csv = node_dir / 'profile_export_aiperf.csv'
        if profile_csv.exists():
            signals = extract_latency_metrics(profile_csv)
            if signals:
                return signals
    return {}
````

## Construção do Resumo por Experimento

Combina, em uma única linha por experimento, os sinais de rede (`aggregate_network_signals`) e de
latência (`aggregate_latency_signals`): throughput médio e de pico, fração de tempo ativo acima de
`COMM_THRESHOLD`, volume total transferido e burstiness (CV = desvio-padrão/média).

Em seguida, para cada grupo `(cluster, seqlen)`, calcula o *overhead* de latência de cada
configuração em relação à baseline N1 (execução em um único nó, sem comunicação inter-nó):

```
overhead_pct = 100 * (latência_multi_nó - latência_N1) / latência_N1
```

Esse percentual é o que permite responder, de forma direta, "quanto a paralelização multi-nó
piora a latência frente a rodar tudo em um nó só?" — a métrica central para testar a hipótese de
gargalo de comunicação. O resultado é salvo em `data_processed/summary_network.csv`.


````python
def aggregate_network_signals(exp_name):
    tp = network_ts_df[network_ts_df['experiment_name'] == exp_name]['total_kB_s'].dropna().values
    if len(tp) == 0:
        return {}
    return {
        'total_comm_avg_kB_s': float(np.mean(tp)),
        'total_comm_peak_kB_s': float(np.max(tp)),
        'active_comm_fraction': float(np.mean(tp > COMM_THRESHOLD)),
        'total_volume_GB': float(np.sum(tp) / 1024 / 1024),
        'burstiness_cv': float(np.std(tp) / np.mean(tp)) if np.mean(tp) > 0 else None,
    }


all_signals = []
for exp_path in experiment_paths:
    meta = parse_exp_name(exp_path.name)
    meta['seqlen'] = workload_for(exp_path.name)  # override folder-name suffix with the corrected label
    comm = aggregate_network_signals(exp_path.name)
    latency = aggregate_latency_signals(exp_path)
    all_signals.append({**meta, **comm, **latency})

summary_df = pd.DataFrame(all_signals)

overhead_rows = []
for (cluster, seqlen), group in summary_df.groupby(['cluster', 'seqlen']):
    n1 = group[group['cluster_size'] == 1]
    baseline_ttft = n1.iloc[0].get('TTFT_avg_ms') if len(n1) > 0 else None
    baseline_inter = n1.iloc[0].get('inter_token_avg_ms') if len(n1) > 0 else None
    for _, row in group.iterrows():
        row_copy = row.copy()
        if baseline_ttft and pd.notna(baseline_ttft) and pd.notna(row.get('TTFT_avg_ms')):
            row_copy['TTFT_overhead_pct'] = 100 * (row['TTFT_avg_ms'] - baseline_ttft) / baseline_ttft
        if baseline_inter and pd.notna(baseline_inter) and pd.notna(row.get('inter_token_avg_ms')):
            row_copy['inter_token_overhead_pct'] = 100 * (row['inter_token_avg_ms'] - baseline_inter) / baseline_inter
        overhead_rows.append(row_copy)

summary_df = pd.DataFrame(overhead_rows)
summary_df.to_csv(OUTPUT_DIR / 'summary_network.csv', index=False)
print(f'Summary: {summary_df.shape}')
summary_df[['exp_name', 'cluster_size', 'parallelism', 'seqlen', 'TTFT_avg_ms', 'total_comm_avg_kB_s']].head(10)
````

## Throughput de Rede

Aqui comparamos os experimentos entre si em termos de throughput médio de rede (RX + TX).
O gráfico de barras ordena os experimentos pelo throughput médio agregado, em escala logarítmica
para acomodar a diferença de ordens de grandeza entre "No Comm" e Tensor Parallel. O heatmap
detalha a contribuição de cada nó individualmente, permitindo ver se o tráfego se concentra em
um nó específico (por exemplo, o primeiro estágio de um pipeline) ou se é distribuído de forma
mais uniforme (como esperado em all-reduce de Tensor Parallel).

É possível ver que os experimentos que usam Tensor Parallel utilizam muito mais a rede que os
outros — resultado esperado, já que TP exige sincronização (all-reduce/all-gather) a cada camada
do modelo, enquanto Pipeline Parallel só troca ativações nas fronteiras entre estágios.


````python
# Average network throughput by experiment — Short and Long combined in one figure

# "none" (sem comunicação inter-nó) fica de fora: throughput de rede não é uma
# métrica relevante para um experimento que não gera tráfego entre nós.
comm_ts_df = network_ts_df[network_ts_df['experiment_name'].apply(get_experiment_type) != 'none']

net_agg = comm_ts_df.groupby('experiment_name').agg(
    rx_kB_s=('rx_kB_s', 'mean'),
    tx_kB_s=('tx_kB_s', 'mean'),
).reset_index()

net_agg['experiment_avg_kB_s'] = net_agg['rx_kB_s'] + net_agg['tx_kB_s']
net_agg['experiment_label'] = net_agg['experiment_name'].apply(format_experiment_name)
net_agg['workload'] = net_agg['experiment_name'].apply(workload_for)

subset = net_agg.sort_values('experiment_avg_kB_s').reset_index(drop=True)

colors = [PARALLELISM_COLORS[get_experiment_type(e)] for e in subset['experiment_name']]
hatches = [WORKLOAD_HATCH[w] for w in subset['workload']]

fig, ax = plt.subplots(figsize=bar_figsize_for(len(subset)))
ax.bar(range(len(subset)), subset['experiment_avg_kB_s'], color=colors, hatch=hatches,
       edgecolor='black', linewidth=1.0)
ax.set_xticks(range(len(subset)))
ax.set_xticklabels(subset['experiment_label'], rotation=45, ha='right', fontsize=10)
ax.set_yscale('log')
ax.set_ylabel('Average Throughput (kB/s, log scale)')
ax.grid(axis='y', linestyle='--', alpha=0.6, which='both')

legend1 = ax.legend(handles=parallelism_legend_patches(subset['experiment_name']), title='Parallelism', loc='upper left')
ax.add_artist(legend1)
ax.legend(handles=WORKLOAD_HATCH_PATCHES, title='Workload', loc='lower right')

fig.suptitle('Average Network Throughput by Experiment')
plt.tight_layout()
save_fig(fig, 'network_throughput_by_experiment.png')
````

````python
# Average network throughput per node (heatmap) — Short and Long combined in one figure

pivot_source = (
    network_ts_df
    .assign(
        experiment_label=lambda d: d['experiment_name'].apply(format_experiment_name),
        node_label=lambda d: d['node_name'].apply(format_node_name),
    )
)

pivot = pivot_source.groupby(['experiment_label', 'node_label'])['total_kB_s'].mean().unstack()

fig_w = max(6.0, 1.6 * len(pivot.columns) + 3.0)
fig_h = max(3.0, 0.8 * len(pivot) + 2.0)
fig, ax = plt.subplots(figsize=(fig_w, fig_h))

sns.heatmap(
    pivot,
    annot=True,
    fmt='.1f',
    cmap='viridis',
    cbar_kws={'label': 'Average Throughput (kB/s)'},
    ax=ax,
)

ax.set_xlabel('Node')
ax.set_ylabel('Experiment')

fig.suptitle('Average Network Throughput per Node')
plt.tight_layout()
save_fig(fig, 'network_throughput_per_node.png')
````

## Throughput ao Longo do Tempo

No pipeline parallelism, o tráfego concentra-se em fases definidas (troca de ativações entre
estágios), produzindo um perfil mais regular e com pausas visíveis entre picos — cada nó só
comunica quando recebe ou repassa uma ativação. O tensor parallelism apresenta comunicação
distribuída ao longo de toda a execução (all-reduce/all-gather a cada camada), gerando maior
irregularidade e picos mais pronunciados e frequentes.

Antes de plotar, `trim_idle_tails_and_normalize_time` remove as pontas ociosas do trace (tempo de
setup do experimento antes da primeira requisição e de encerramento após a última), para que o
eixo do tempo reflita apenas a janela de execução ativa.


````python
# Network throughput over time — one grid with all experiments, Short and Long combined

exp_list = ordered_experiments

n_exp = len(exp_list)
n_cols = min(3, n_exp)
n_rows = math.ceil(n_exp / n_cols)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(5.5 * n_cols, 3.8 * n_rows), squeeze=False)
axes_flat = axes.flatten()

for ax, exp_name in zip(axes_flat, exp_list):
    exp_df = network_ts_df[network_ts_df['experiment_name'] == exp_name].copy()
    exp_df = trim_idle_tails_and_normalize_time(exp_df)
    for node in sorted(exp_df['node_name'].unique()):
        node_df = exp_df[exp_df['node_name'] == node].sort_values('t')
        ax.plot(node_df['t'], node_df['total_kB_s'], linewidth=1.5, label=format_node_name(node))
    ax.set_title(f"{format_experiment_name(exp_name)} [{get_experiment_type(exp_name)}]", fontsize=10)
    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Throughput (kB/s)')
    ax.grid(True, linestyle='--', alpha=0.4)
    ax.legend(fontsize=8, title='Node', title_fontsize=9)

for ax in axes_flat[n_exp:]:
    ax.set_visible(False)

fig.suptitle('Network Throughput Over Time — All Experiments')
plt.tight_layout()
save_fig(fig, 'all_experiments_network_throughput.png')
````

### Volume Cumulativo de Comunicação

Revela *quando* o tráfego ocorre: crescimento linear ao longo do eixo do tempo indica comunicação
distribuída uniformemente durante a execução (típico de TP, com sincronização a cada camada);
crescimento em degraus indica rajadas pontuais concentradas em poucos instantes (típico de PP,
onde a curva sobe apenas nos momentos de troca de ativação entre estágios e fica plana no resto
do tempo, enquanto o nó apenas computa).


````python
# Cumulative communication volume — one grid with all experiments, Short and Long combined

exp_list = sorted(ordered_experiments, key=get_experiment_type)

n_exp = len(exp_list)
n_cols = min(3, n_exp)
n_rows = math.ceil(n_exp / n_cols)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(5.5 * n_cols, 3.8 * n_rows), squeeze=False)
axes_flat = axes.flatten()

for ax, exp_name in zip(axes_flat, exp_list):
    exp_df = network_ts_df[network_ts_df['experiment_name'] == exp_name].copy()
    per_second = second_level_node_df(exp_df).sort_values(['node_name', 'timestamp'])
    per_second['t'] = (per_second['timestamp'] - per_second['timestamp'].min()).dt.total_seconds()
    per_second['cumulative_GB'] = (
        per_second.groupby('node_name')['total_kB_s'].cumsum() / (1024 * 1024)
    )
    for node in sorted(per_second['node_name'].unique()):
        node_df = per_second[per_second['node_name'] == node]
        ax.plot(node_df['t'], node_df['cumulative_GB'], linewidth=2.0, label=format_node_name(node))
    ax.set_title(format_experiment_name(exp_name), fontsize=10)
    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Cumulative Volume (GB)')
    ax.grid(True, alpha=0.3)
    ax.legend(title='Node', fontsize=8, title_fontsize=9)

for ax in axes_flat[n_exp:]:
    ax.set_visible(False)

fig.suptitle('Cumulative Communication Volume — All Experiments')
plt.tight_layout()
save_fig(fig, 'cumulative_communication_grid.png')
````

## Volume de Comunicação e Burstiness

Volume total de comunicação por experimento (soma de `total_kB_s` ao longo de todo o trace,
convertida para GB) e burstiness (coeficiente de variação, CV = desvio-padrão/média, das amostras
de throughput). Valores elevados de CV indicam comunicação concentrada em picos curtos e intensos
em vez de distribuída de forma uniforme — o que é relevante porque tráfego "em rajada" tende a
saturar buffers e switches momentaneamente, afetando a latência de cauda mesmo quando o volume
médio é moderado.

Referências: JAGERMAN & MELAMED (1994); LUANGSOMBOON et al. (2024) arXiv:2401.00329.


````python
# Communication volume per experiment — Short and Long combined in one figure

# "none" (sem comunicação inter-nó) fica de fora: volume de comunicação não é uma
# métrica relevante para um experimento que não gera tráfego entre nós.
comm_ts_df = network_ts_df[network_ts_df['experiment_name'].apply(get_experiment_type) != 'none']

experiment_comm = comm_ts_df.groupby('experiment_name')['total_kB_s'].sum().reset_index()
experiment_comm['total_GB'] = experiment_comm['total_kB_s'] / (1024 * 1024)
experiment_comm['experiment_label'] = experiment_comm['experiment_name'].apply(format_experiment_name)
experiment_comm['workload'] = experiment_comm['experiment_name'].apply(workload_for)

subset = experiment_comm.sort_values('total_GB').reset_index(drop=True)

colors = [PARALLELISM_COLORS[get_experiment_type(e)] for e in subset['experiment_name']]
hatches = [WORKLOAD_HATCH[w] for w in subset['workload']]

fig, ax = plt.subplots(figsize=bar_figsize_for(len(subset)))
ax.bar(range(len(subset)), subset['total_GB'], color=colors, hatch=hatches, edgecolor='black', linewidth=1.0)
ax.set_xticks(range(len(subset)))
ax.set_xticklabels(subset['experiment_label'], rotation=45, ha='right', fontsize=10)
ax.set_yscale('log')
ax.set_ylabel('Total Network Communication (GB, log scale)')
ax.grid(axis='y', linestyle='--', alpha=0.6, which='both')

legend1 = ax.legend(handles=parallelism_legend_patches(subset['experiment_name']), title='Parallelism', loc='upper left')
ax.add_artist(legend1)
ax.legend(handles=WORKLOAD_HATCH_PATCHES, title='Workload', loc='lower right')

fig.suptitle('Communication Volume per Experiment')
plt.tight_layout()
save_fig(fig, 'communication_volume_bar.png')
````

### Eficiência de Comunicação por Nó

Normalizamos o volume total pelo número de nós, obtendo a comunicação média gerada por nó
(`GB_per_node`). Isso permite comparar experimentos com diferentes tamanhos de cluster em pé de
igualdade: sem essa normalização, um experimento N4 pareceria "usar mais rede" apenas por ter mais
nós somando tráfego, mesmo que cada nó individualmente comunique pouco. O eixo x tem leve jitter
para que pontos sobrepostos no mesmo tamanho de cluster fiquem visíveis.


````python
from matplotlib.lines import Line2D

comm_df = (
    network_ts_df.groupby(['experiment_name', 'node_name'])['total_kB_s'].sum().reset_index()
    .groupby('experiment_name').agg(
        total_kB_s=('total_kB_s', 'sum'),
        n_nodes=('node_name', 'nunique'),
    ).reset_index()
)
comm_df['total_GB'] = comm_df['total_kB_s'] / (1024 * 1024)
comm_df['GB_per_node'] = comm_df['total_GB'] / comm_df['n_nodes']
comm_df['ptype'] = comm_df['experiment_name'].apply(get_experiment_type)
comm_df['workload'] = comm_df['experiment_name'].apply(workload_for)
comm_df['experiment_label'] = comm_df['experiment_name'].apply(format_experiment_name)

rng = np.random.default_rng(42)
jitter_y = comm_df['GB_per_node'].std() * 0.03
comm_df['n_nodes_jitter'] = comm_df['n_nodes'] + rng.uniform(-0.18, 0.18, len(comm_df))
comm_df['gb_per_node_jitter'] = comm_df['GB_per_node'] + rng.uniform(-jitter_y, jitter_y, len(comm_df))

fig, ax = plt.subplots(figsize=(12, 7))
for workload in ['Short', 'Long']:
    subset_wl = comm_df[comm_df['workload'] == workload]
    for ptype in sorted(subset_wl['ptype'].unique(), key=lambda t: TYPE_ORDER.get(t, 99)):
        subset = subset_wl[subset_wl['ptype'] == ptype]
        ax.scatter(subset['n_nodes_jitter'], subset['gb_per_node_jitter'],
                   c=PARALLELISM_COLORS[ptype], marker=WORKLOAD_MARKERS[workload], s=140, alpha=0.8,
                   edgecolors='black', linewidth=0.8)

for _, row in comm_df.iterrows():
    ax.annotate(row['experiment_label'], (row['n_nodes_jitter'], row['gb_per_node_jitter']),
                textcoords='offset points', xytext=(6, 4), fontsize=9)

ax.set_xlabel('Number of Nodes')
ax.set_ylabel('Communication Volume per Node (GB)')
ax.set_xticks([1, 2, 3, 4])
ax.grid(True, alpha=0.3)

legend1 = ax.legend(handles=parallelism_legend_patches(comm_df['experiment_name']), title='Parallelism', loc='upper left')
ax.add_artist(legend1)
workload_handles = [Line2D([0], [0], marker=WORKLOAD_MARKERS[wl], color='gray', linestyle='None',
                            markersize=10, label=wl) for wl in ['Short', 'Long']]
ax.legend(handles=workload_handles, title='Workload', loc='upper right')

fig.suptitle('Communication Volume per Node vs Number of Nodes')
plt.tight_layout()
save_fig(fig, 'communication_volume_scaling_scatter.png')
````

````python
# "none" (sem comunicação inter-nó) fica de fora: burstiness de rede não faz sentido
# para um experimento que não gera tráfego entre nós.
comm_ts_df = network_ts_df[network_ts_df['experiment_name'].apply(get_experiment_type) != 'none']

burstiness_df = (
    comm_ts_df.groupby('experiment_name')['total_kB_s']
    .agg(lambda x: x.std() / (x.mean() + 1e-9))
    .reset_index()
)
burstiness_df.columns = ['experiment_name', 'burstiness_cv']
burstiness_df['experiment_label'] = burstiness_df['experiment_name'].apply(format_experiment_name)
burstiness_df['workload'] = burstiness_df['experiment_name'].apply(workload_for)

node_stats = (
    comm_ts_df.groupby(['experiment_name', 'node_name'])
    .agg(mean_kB_s=('total_kB_s', 'mean'), std_kB_s=('total_kB_s', 'std'))
    .reset_index()
)
node_stats['cv'] = node_stats['std_kB_s'] / (node_stats['mean_kB_s'] + 1e-9)
node_cv_df = node_stats.groupby('experiment_name')['cv'].mean().reset_index()
node_cv_df.columns = ['experiment_name', 'node_cv']
node_cv_df['experiment_label'] = node_cv_df['experiment_name'].apply(format_experiment_name)
node_cv_df['workload'] = node_cv_df['experiment_name'].apply(workload_for)

burst_subset = burstiness_df.sort_values('burstiness_cv').reset_index(drop=True)
node_subset = node_cv_df.sort_values('node_cv').reset_index(drop=True)

panel_w = bar_figsize_for(max(len(burst_subset), len(node_subset)))[0]
fig, axes = plt.subplots(1, 2, figsize=(panel_w * 2, 7))

ax = axes[0]
colors = [PARALLELISM_COLORS[get_experiment_type(e)] for e in burst_subset['experiment_name']]
hatches = [WORKLOAD_HATCH[w] for w in burst_subset['workload']]
ax.bar(range(len(burst_subset)), burst_subset['burstiness_cv'], color=colors, hatch=hatches,
       edgecolor='black', linewidth=1.0)
ax.set_xticks(range(len(burst_subset)))
ax.set_xticklabels(burst_subset['experiment_label'], rotation=45, ha='right', fontsize=10)
ax.set_ylabel('Burstiness (std/mean)')
ax.set_title('Per-Experiment Burstiness')
ax.grid(axis='y', linestyle='--', alpha=0.6)
ax.legend(handles=parallelism_legend_patches(burst_subset['experiment_name']), title='Parallelism')

ax = axes[1]
colors = [PARALLELISM_COLORS[get_experiment_type(e)] for e in node_subset['experiment_name']]
hatches = [WORKLOAD_HATCH[w] for w in node_subset['workload']]
ax.bar(range(len(node_subset)), node_subset['node_cv'], color=colors, hatch=hatches,
       edgecolor='black', linewidth=1.0)
ax.set_xticks(range(len(node_subset)))
ax.set_xticklabels(node_subset['experiment_label'], rotation=45, ha='right', fontsize=10)
ax.set_ylabel('Coefficient of Variation')
ax.set_title('Per-Node Coefficient of Variation')
ax.grid(axis='y', linestyle='--', alpha=0.6)
ax.legend(handles=parallelism_legend_patches(node_subset['experiment_name']), title='Parallelism')

fig.legend(handles=WORKLOAD_HATCH_PATCHES, title='Workload', loc='upper center', ncol=2, bbox_to_anchor=(0.5, 1.04))
fig.suptitle('Communication Burstiness Analysis')
plt.tight_layout()
save_fig(fig, 'communication_burstiness.png')
````

## Atividade de Comunicação e Overhead

Em vez de medir apenas volume, esta seção mede **tempo**: por quanto da execução cada experimento
esteve efetivamente comunicando dados pela rede (acima do limiar `COMM_THRESHOLD` de 10 kB/s),
versus quanto tempo ficou em silêncio de rede (presumivelmente computando). Dois experimentos
podem ter o mesmo volume total, mas se um concentra tudo em uma fração pequena do tempo (alta
intensidade, baixa duração) e o outro comunica de forma constante (baixa intensidade, longa
duração), o impacto na latência pode ser bem diferente — daí a necessidade de olhar volume e
tempo de atividade separadamente.


````python
# "none" (sem comunicação inter-nó) fica de fora: tempo ativo em rede não é uma métrica
# relevante para um experimento que não gera tráfego entre nós.
comm_ts_df = network_ts_df[network_ts_df['experiment_name'].apply(get_experiment_type) != 'none']

per_second = (
    comm_ts_df.assign(timestamp=pd.to_datetime(comm_ts_df['timestamp']))
    .groupby(['experiment_name', pd.Grouper(key='timestamp', freq='1s')])['total_kB_s']
    .sum().reset_index()
)
comm_time_df = (
    per_second.groupby('experiment_name').agg(
        comm_fraction=('total_kB_s', lambda x: (x > COMM_THRESHOLD).mean()),
        comm_active_seconds=('total_kB_s', lambda x: (x > COMM_THRESHOLD).sum()),
    ).reset_index()
)
comm_time_df['experiment_label'] = comm_time_df['experiment_name'].apply(format_experiment_name)
comm_time_df['workload'] = comm_time_df['experiment_name'].apply(workload_for)

subset = comm_time_df.reset_index(drop=True)

colors = [PARALLELISM_COLORS[get_experiment_type(e)] for e in subset['experiment_name']]
hatches = [WORKLOAD_HATCH[w] for w in subset['workload']]

panel_w = bar_figsize_for(len(subset))[0]
fig, axes = plt.subplots(1, 2, figsize=(panel_w * 2, 7))

ax = axes[0]
ax.bar(range(len(subset)), subset['comm_fraction'], color=colors, hatch=hatches,
       edgecolor='black', linewidth=1.0)
ax.set_xticks(range(len(subset)))
ax.set_xticklabels(subset['experiment_label'], rotation=45, ha='right', fontsize=10)
ax.set_ylabel('Fraction of Time in Communication')
ax.set_title('Communication Time Fraction')
ax.grid(axis='y', linestyle='--', alpha=0.6)
ax.legend(handles=parallelism_legend_patches(subset['experiment_name']), title='Parallelism')

ax = axes[1]
ax.bar(range(len(subset)), subset['comm_active_seconds'], color=colors, hatch=hatches,
       edgecolor='black', linewidth=1.0)
ax.set_xticks(range(len(subset)))
ax.set_xticklabels(subset['experiment_label'], rotation=45, ha='right', fontsize=10)
ax.set_ylabel('Estimated Communication Time (s)')
ax.set_title('Active Communication Time')
ax.grid(axis='y', linestyle='--', alpha=0.6)
ax.legend(handles=parallelism_legend_patches(subset['experiment_name']), title='Parallelism')

fig.legend(handles=WORKLOAD_HATCH_PATCHES, title='Workload', loc='upper center', ncol=2, bbox_to_anchor=(0.5, 1.04))
fig.suptitle('Communication Activity and Overhead')
plt.tight_layout()
save_fig(fig, 'communication_activity.png')
````

## Topologia de Rede

Distribuição relativa do tráfego por nó, normalizada por experimento (cada linha do heatmap soma
1.0). Identifica desbalanceamento entre nós dentro de cada estratégia de paralelismo: em Pipeline
Parallel, é esperado que os estágios intermediários concentrem mais tráfego que os extremos, já
que eles recebem ativações do estágio anterior e enviam para o seguinte; em Tensor Parallel, o
tráfego deveria se distribuir de forma mais equilibrada entre os nós, já que todos participam
igualmente de cada operação de all-reduce/all-gather.


````python
topology_data = (
    network_ts_df
    .assign(
        experiment_label=lambda d: d['experiment_name'].apply(format_experiment_name),
        node_label=lambda d: d['node_name'].apply(format_node_name),
    )
)

topology_raw = topology_data.groupby(['experiment_label', 'node_label'])['total_kB_s'].mean().unstack()
topology_norm = topology_raw.div(topology_raw.sum(axis=1), axis=0)

fig_w = max(6.0, 1.6 * len(topology_norm.columns) + 3.0)
fig_h = max(4.0, 0.8 * len(topology_norm) + 2.0)
fig, ax = plt.subplots(figsize=(fig_w, fig_h))

sns.heatmap(
    topology_norm, annot=True, fmt='.2f', cmap='magma',
    cbar_kws={'label': 'Fraction of Total Communication'}, ax=ax,
)
ax.set_xlabel('Node')
ax.set_ylabel('Experiment')

fig.suptitle('Relative Communication Distribution Across Nodes')
plt.tight_layout()
save_fig(fig, 'network_topology_heatmap_normalized.png')
````

## Análise de Gargalo

Testa se a intensidade de comunicação se correlaciona com a latência de inferência, e se a
latência se degrada com o tamanho do cluster e a estratégia de paralelismo escolhida.

**Hipótese**: maior volume/intensidade de comunicação se correlaciona com maior latência.
Curvas de escalonamento não planas (latência crescendo com o número de nós) indicam um gargalo de
comunicação — ou seja, o custo de sincronizar dados entre nós passa a dominar sobre o ganho de
paralelizar o cômputo.

As células a seguir atacam essa hipótese de três ângulos complementares:
1. **Dispersão latência × comunicação**: TTFT/latência entre tokens versus volume e throughput de
   rede, separado por estratégia de paralelismo.
2. **Curvas de escalonamento**: como TTFT e latência entre tokens evoluem com o número de nós,
   por estratégia de paralelismo (sequências Short apenas).
3. **Latência de cauda (p99)**: se burstiness e fração de tempo em comunicação afetam
   desproporcionalmente os percentis altos de latência, que é onde gargalos de rede costumam se
   manifestar de forma mais visível do que na média.


````python
valid_ttft = summary_df.dropna(subset=['TTFT_avg_ms', 'total_volume_GB', 'parallelism', 'seqlen'])
valid_itl = summary_df.dropna(subset=['inter_token_avg_ms', 'total_comm_avg_kB_s', 'parallelism', 'seqlen'])

fig, axes = plt.subplots(1, 2, figsize=(14, 7))

ax = axes[0]
for par in valid_ttft['parallelism'].unique():
    for workload in ['Short', 'Long']:
        subset = valid_ttft[(valid_ttft['parallelism'] == par) & (valid_ttft['seqlen'] == workload)]
        if subset.empty:
            continue
        ax.scatter(subset['total_volume_GB'], subset['TTFT_avg_ms'],
                   color=PARALLELISM_COLORS[par.lower()], marker=WORKLOAD_MARKERS[workload],
                   label=f'{par} ({workload})', s=120, alpha=0.8, edgecolors='black', linewidth=0.6)
ax.set_xlabel('Total Communication Volume (GB)')
ax.set_ylabel('TTFT (ms)')
ax.set_title('TTFT vs Communication Volume')
ax.legend(title='Parallelism (Workload)')
ax.grid(True, alpha=0.3)

ax = axes[1]
for par in valid_itl['parallelism'].unique():
    for workload in ['Short', 'Long']:
        subset = valid_itl[(valid_itl['parallelism'] == par) & (valid_itl['seqlen'] == workload)]
        if subset.empty:
            continue
        ax.scatter(subset['total_comm_avg_kB_s'], subset['inter_token_avg_ms'],
                   color=PARALLELISM_COLORS[par.lower()], marker=WORKLOAD_MARKERS[workload],
                   label=f'{par} ({workload})', s=120, alpha=0.8, edgecolors='black', linewidth=0.6)
ax.set_xlabel('Avg Communication Throughput (kB/s)')
ax.set_ylabel('Inter-Token Latency (ms)')
ax.set_title('Inter-Token Latency vs Communication Throughput')
ax.legend(title='Parallelism (Workload)')
ax.grid(True, alpha=0.3)

fig.suptitle('Latency vs Communication')
plt.tight_layout()
save_fig(fig, 'latency_vs_communication.png')
````
